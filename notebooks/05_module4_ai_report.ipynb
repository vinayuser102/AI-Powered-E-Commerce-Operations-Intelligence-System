```python
import sys
from pathlib import Path

# Handle dynamic root path
ROOT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))

from core.ai_reporter import AIReporterEngine

# 1. Initialize AI Reporter Engine
reporter = AIReporterEngine(output_dir=str(ROOT_DIR / "outputs"))

# 2. Paths to Processed Data
rfm_path = ROOT_DIR / "data" / "processed" / "rfm_segments.csv"
forecast_path = ROOT_DIR / "data" / "processed" / "demand_forecast.csv"
anomalies_path = ROOT_DIR / "data" / "processed" / "seller_anomalies.csv"

# 3. Generate Reports via Groq LLM
summary_text, supplier_text = reporter.generate_reports(
    rfm_path=str(rfm_path),
    forecast_path=str(forecast_path),
    anomalies_path=str(anomalies_path)
)

print("=== AI GENERATED EXECUTIVE SUMMARY ===")
print(summary_text[:300] + "...\n")

# 4. Dispatch Email via Brevo SMTP Relay
reporter.send_report_email(
    summary_text=summary_text,
    supplier_text=supplier_text,
    recipient_email="vinaygod100@gmail.com",
    dashboard_url="https://your-app.streamlit.app"
)

print("\nModule 4 Complete! Output saved & email dispatched.")
```
