```python
import sys
from pathlib import Path
import pandas as pd

# Handle dynamic root path
ROOT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))

from ml_engine.anomaly_detector import SellerAnomalyEngine

# 1. Initialize Anomaly Engine
engine = SellerAnomalyEngine(
    model_dir=str(ROOT_DIR / "models"), 
    output_dir=str(ROOT_DIR / "outputs"),
    contamination=0.05
)

# 2. Load Data & Aggregate Seller Metrics
df_master = pd.read_csv(ROOT_DIR / "data" / "processed" / "master_data.csv")
seller_stats = engine.extract_seller_features(df_master)

# 3. Detect Anomalies & Save Visualization
seller_stats = engine.detect_anomalies(seller_stats)
engine.save_anomaly_plot(seller_stats)

# 4. Save Processed Results
seller_stats.to_csv(ROOT_DIR / "data" / "processed" / "seller_anomalies.csv", index=False)

anomalies = seller_stats[seller_stats['anomaly'] == -1]

print("=== Module 3 Summary ===")
print(f"Total sellers analysed: {len(seller_stats):,}")
print(f"Anomalous sellers flagged: {len(anomalies):,}")
print(f"Anomaly rate: {len(anomalies)/len(seller_stats)*100:.1f}%")
print(f"\nWorst delay anomaly: {anomalies['avg_delay'].max():.1f} days")
print(f"Highest revenue anomaly: R${anomalies['total_revenue'].max():,.2f}")
print("\nModule 3 Complete! Results saved to seller_anomalies.csv")
```
